```python
!pip -q install opencv-python-headless

import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

print("OpenCV version:", cv2.__version__)
```

```python
uploaded = files.upload()

image_name = next(iter(uploaded.keys()))

image = cv2.imread(image_name)

if image is None:
    raise ValueError("The uploaded file is not a valid image.")

# Convert color image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Image loaded successfully:", image_name)
print("Original image shape:", image.shape)
print("Grayscale image shape:", gray.shape)
```

```python
# Display the original grayscale image
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

# Generate and display histogram
plt.subplot(1, 2, 2)
plt.hist(gray.ravel(), bins=256, range=[0, 256])
plt.title("Histogram of Original Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# Find minimum and maximum pixel intensity
min_val = np.min(gray)
max_val = np.max(gray)

# Apply contrast stretching
contrast_stretched = cv2.normalize(
    gray,
    None,
    0,
    255,
    cv2.NORM_MINMAX
)

print("Original minimum intensity:", min_val)
print("Original maximum intensity:", max_val)
print("After contrast stretching:")
print("Minimum intensity:", np.min(contrast_stretched))
print("Maximum intensity:", np.max(contrast_stretched))

# Display original and contrast-stretched images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(contrast_stretched, cmap="gray")
plt.title("Contrast Stretched Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Apply Histogram Equalization
hist_equalized = cv2.equalizeHist(gray)

# Display original and histogram-equalized images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(hist_equalized, cmap="gray")
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2

import numpy as np

import matplotlib.pyplot as plt

clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))

clahe_img = clahe.apply(gray)

plt.figure(figsize=(5,5))

plt.imshow(clahe_img, cmap='gray')

plt.title("CLAHE Enhanced, Image")

plt.axis('off')

plt.show()
```

```python
import cv2

import matplotlib.pyplot as plt

import numpy as np

fig, axs = plt.subplots(1, 3, figsize=(15,4))

axs[0].hist(gray.ravel(), bins=256, range=[0,256], color='gray')

axs[0].set_title("Original Histogram")

axs[1].hist(hist_equalized.ravel(), bins=256, range=[0,256], color="blue")

axs[1].set_title("Equalized Histogram")

axs[2].hist(clahe_img.ravel(), bins=256, range=[0,256], color='green')

axs[2].set_title("CLAHE Histogram")

plt.tight_layout()

plt.show()
```
