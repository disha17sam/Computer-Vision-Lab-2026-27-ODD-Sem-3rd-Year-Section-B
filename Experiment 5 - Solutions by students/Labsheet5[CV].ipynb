```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from skimage.feature import hog
from skimage import exposure
from google.colab import files
```

```python
uploaded = files.upload()

filename = next(iter(uploaded))
img = cv2.imread(filename)

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8,6))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
sift = cv2.SIFT_create()

keypoints, descriptors = sift.detectAndCompute(gray, None)

print("Number of Keypoints:", len(keypoints))
print("Descriptor Shape:", descriptors.shape)
```

```python
sift_image = cv2.drawKeypoints(
    img_rgb,
    keypoints,
    None,
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)

plt.figure(figsize=(10,7))
plt.imshow(sift_image)
plt.title("SIFT Keypoints")
plt.axis("off")
plt.show()
```

```python
# Resize image for consistent HOG extraction
hog_image = cv2.resize(gray, (128, 128))

features, hog_visual = hog(
    hog_image,
    orientations=9,
    pixels_per_cell=(8, 8),
    cells_per_block=(2, 2),
    visualize=True
)

print("HOG Feature Vector Length:", len(features))
```

```python
hog_visual = exposure.rescale_intensity(
    hog_visual,
    in_range=(0, 10)
)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(hog_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(hog_visual, cmap="gray")
plt.title("HOG Visualization")
plt.axis("off")

plt.show()
```

```python
uploaded2 = files.upload()

filename2 = next(iter(uploaded2))
img2 = cv2.imread(filename2)

gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8,6))
plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.title("Second Image")
plt.axis("off")
plt.show()
```

```python
keypoints2, descriptors2 = sift.detectAndCompute(gray2, None)

print("Image 2 Keypoints:", len(keypoints2))
```

```python
bf = cv2.BFMatcher()

matches = bf.knnMatch(descriptors, descriptors2, k=2)

# Lowe's ratio test
good_matches = []

for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good_matches.append(m)

print("Good Matches:", len(good_matches))
```

```python
match_image = cv2.drawMatches(
    img,
    keypoints,
    img2,
    keypoints2,
    good_matches,
    None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

plt.figure(figsize=(15,8))
plt.imshow(cv2.cvtColor(match_image, cv2.COLOR_BGR2RGB))
plt.title("SIFT Image Matching")
plt.axis("off")
plt.show()
```
