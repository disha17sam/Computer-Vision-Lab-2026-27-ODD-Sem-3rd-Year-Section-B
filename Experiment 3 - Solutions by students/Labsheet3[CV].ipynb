```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('brushes.jpeg')

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Image loaded successfully!")
print("Image Shape:", gray.shape)
```

```python
plt.figure(figsize=(8, 6))

plt.imshow(gray, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.show()
```

```python
gaussian = cv2.GaussianBlur(gray, (5, 5), 0)

plt.figure(figsize=(8, 6))

plt.imshow(gaussian, cmap='gray')
plt.title('Gaussian Blur')
plt.axis('off')

plt.show()
```

```python
median = cv2.medianBlur(gray, 5)

plt.figure(figsize=(8, 6))

plt.imshow(median, cmap='gray')
plt.title('Median Filter')
plt.axis('off')

plt.show()
```

```python
average = cv2.blur(gray, (5, 5))

plt.figure(figsize=(8, 6))

plt.imshow(average, cmap='gray')
plt.title('Average (Mean) Filter')
plt.axis('off')

plt.show()
```

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)

laplacian = np.uint8(np.absolute(laplacian))

plt.figure(figsize=(8, 6))

plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Filter')
plt.axis('off')

plt.show()
```

```python
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)

sobel_x = np.uint8(np.absolute(sobel_x))

plt.figure(figsize=(8, 6))

plt.imshow(sobel_x, cmap='gray')
plt.title('Sobel X - Horizontal Edges')
plt.axis('off')

plt.show()
```

```python
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

sobel_y = np.uint8(np.absolute(sobel_y))

plt.figure(figsize=(8, 6))

plt.imshow(sobel_y, cmap='gray')
plt.title('Sobel Y - Vertical Edges')
plt.axis('off')

plt.show()
```

```python
sobel_combined = cv2.magnitude(
    sobel_x.astype(np.float32),
    sobel_y.astype(np.float32)
)

sobel_combined = np.uint8(sobel_combined)

plt.figure(figsize=(8, 6))

plt.imshow(sobel_combined, cmap='gray')
plt.title('Combined Sobel Edge Detection')
plt.axis('off')

plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(gray, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(gaussian, cmap='gray')
plt.title('Gaussian')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(median, cmap='gray')
plt.title('Median')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(average, cmap='gray')
plt.title('Average')
plt.axis('off')

plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(sobel_x, cmap='gray')
plt.title('Sobel X')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sobel_y, cmap='gray')
plt.title('Sobel Y')
plt.axis('off')

plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(gaussian, cmap='gray')
plt.title('Gaussian Blur')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(median, cmap='gray')
plt.title('Median Filter')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(average, cmap='gray')
plt.title('Average Filter')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
print("FILTER COMPARISON")
print("-----------------")

print("Original Mean Intensity:", np.mean(gray))
print("Gaussian Mean Intensity:", np.mean(gaussian))
print("Median Mean Intensity:", np.mean(median))
print("Average Mean Intensity:", np.mean(average))

print("\nStandard Deviation:")
print("Original:", np.std(gray))
print("Gaussian:", np.std(gaussian))
print("Median:", np.std(median))
print("Average:", np.std(average))
```

```python
print("OBSERVATIONS")
print("------------")

print("1. Gaussian filtering smooths the image and reduces noise while preserving important structures.")
print("2. Median filtering is effective for removing impulse or salt-and-pepper noise.")
print("3. Average filtering provides basic smoothing but can blur edges more than Gaussian filtering.")
print("4. Laplacian filtering emphasizes edges and fine image details.")
print("5. Sobel X detects intensity changes in the horizontal direction.")
print("6. Sobel Y detects intensity changes in the vertical direction.")
print("7. Low-pass filters are mainly useful for smoothing and noise reduction.")
print("8. High-pass filters are useful for edge detection and detail enhancement.")
```
